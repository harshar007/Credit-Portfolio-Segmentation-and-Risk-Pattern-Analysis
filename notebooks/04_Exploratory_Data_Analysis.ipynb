# 04 - Exploratory Data Analysis (EDA)
### Credit Portfolio Segmentation and Risk Pattern Analysis

---

## Exploring Portfolio Patterns Visually
Before fitting any machine learning models, we look at the distributions and correlations across our borrower base:
1. What does the spread of income and loan requests look like?
2. Do higher credit scores correspond to lower default rates?
3. How are financial attributes correlated with one another?


```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)

df = pd.read_csv("../data/processed/cleaned_loans_dataset.csv")
print(f"Portfolio loaded: {len(df):,} borrowers.")
```

---
## 1. Income & Loan Size Distributions
Let's look at how borrower incomes and requested borrowing amounts are shaped.


```
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(df['annual_income'], bins=30, kde=True, ax=axes[0], color='royalblue')
axes[0].set_title("Annual Income Distribution")
axes[0].set_xlabel("Annual Income ($)")
axes[0].set_xlim(0, 200000)

sns.histplot(df['loan_amount'], bins=30, kde=True, ax=axes[1], color='darkorange')
axes[1].set_title("Loan Amount Distribution")
axes[1].set_xlabel("Loan Amount ($)")

plt.tight_layout()
plt.savefig("../outputs/figures/eda_distributions.png", dpi=300)
plt.show()
```

---
## 2. Credit Score vs. Loan Default
Does credit score separate good loans from bad loans? We check with a box plot.


```
plt.figure(figsize=(8, 5))
sns.boxplot(x='Default', y='fico_score', data=df, palette=['#2ecc71', '#e74c3c'])
plt.title("Credit Score (FICO) by Loan Outcome")
plt.xlabel("Default Status (0 = Paid, 1 = Defaulted)")
plt.ylabel("FICO Credit Score")

plt.tight_layout()
plt.savefig("../outputs/figures/eda_fico_vs_default.png", dpi=300)
plt.show()
```

---
## 3. Financial Correlation Matrix
Let's see how our numerical financial variables interact.


```
feature_cols = ['annual_income', 'dti', 'loan_amount', 'fico_score', 'emp_length_years', 'Default']

plt.figure(figsize=(8, 6))
sns.heatmap(df[feature_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation Matrix of Financial Variables")

plt.tight_layout()
plt.savefig("../outputs/figures/eda_correlation_heatmap.png", dpi=300)
plt.show()
```

---
## Visual Insights
- FICO score shows a strong negative association with Default: lower scores carry visibly higher default rates.
- Higher earners naturally qualify for larger loans, but their default rate stays low unless their debt-to-income ratio is also stretched.
