# 04 - Exploratory Data Analysis (EDA)
### Credit Portfolio Segmentation and Risk Pattern Analysis

---

## 📌 Beginner Objective
EDA helps us **visually discover trends and relationships** before running any machine learning:
1. What do borrower incomes and loan amounts look like?
2. Do borrowers with lower FICO credit scores default more often?
3. How does Debt-to-Income (DTI) correlate with risk?
4. Save plots directly to `../outputs/figures/` for our project reports.


```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set visualization style
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)

# Load cleaned data
df = pd.read_csv("../data/processed/cleaned_loans_dataset.csv")
print(f"Dataset ready with {len(df):,} borrowers.")
```

---
## 1. Univariate Distribution: Income and Loan Amount
Let's see how money is distributed across applicants.


```
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot Annual Income (capped at $200k for visual clarity)
sns.histplot(df['annual_income'], bins=30, kde=True, ax=axes[0], color='royalblue')
axes[0].set_title("Distribution of Annual Income")
axes[0].set_xlabel("Annual Income ($)")
axes[0].set_xlim(0, 200000)

# Plot Loan Amount
sns.histplot(df['loan_amount'], bins=30, kde=True, ax=axes[1], color='darkorange')
axes[1].set_title("Distribution of Loan Amount")
axes[1].set_xlabel("Loan Amount ($)")

plt.tight_layout()
plt.savefig("../outputs/figures/eda_distributions.png", dpi=300)
plt.show()
```

---
## 2. Bivariate Analysis: FICO Score vs. Default
Do borrowers with lower FICO scores default more? Let's check with a box plot.


```
plt.figure(figsize=(8, 5))
sns.boxplot(x='Default', y='fico_score', data=df, palette=['#2ecc71', '#e74c3c'])
plt.title("Borrower FICO Score by Default Status")
plt.xlabel("Default Status (0 = Fully Paid, 1 = Defaulted)")
plt.ylabel("FICO Credit Score")

plt.tight_layout()
plt.savefig("../outputs/figures/eda_fico_vs_default.png", dpi=300)
plt.show()
```

---
## 3. Correlation Heatmap
Let's examine how the financial variables correlate with each other.


```
feature_cols = ['annual_income', 'dti', 'loan_amount', 'fico_score', 'emp_length_years', 'Default']

plt.figure(figsize=(8, 6))
sns.heatmap(df[feature_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation Matrix of Financial Variables")

plt.tight_layout()
plt.savefig("../outputs/figures/eda_correlation_heatmap.png", dpi=300)
plt.show()
```

---
## 💡 Beginner Summary & Next Step
- FICO score shows a strong negative relationship with Default: lower scores have significantly higher default occurrences.
- Income is positively correlated with Loan Amount (banks lend more to higher earners).
