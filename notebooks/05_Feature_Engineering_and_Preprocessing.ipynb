## Preprocessing for Unsupervised Clustering
Clustering algorithms rely on Euclidean distances between data points.
- If borrower A earns $80,000 and borrower B earns $50,000, their income difference is 30,000.
- If their debt ratio differs by 5%, that difference is just 5.
- Without standardization, the algorithm would treat income as thousands of times more important than debt burden!

We apply `StandardScaler` to ensure every variable has a mean of 0 and a standard deviation of 1.

> [!IMPORTANT]
> **No Target Leakage:**
> We explicitly exclude the `Default` column from this step. We want the clustering models to group borrowers purely based on their financial traits, without peeking at the repayment outcome.


```
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from pathlib import Path

df = pd.read_csv("../data/processed/cleaned_loans_dataset.csv")

features = ['annual_income', 'dti', 'loan_amount', 'fico_score', 'emp_length_years']
X = df[features].copy()

print("Selected features for portfolio clustering:")
print(X.head())
```

---
## 1. Applying StandardScaler


```
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

df_scaled = pd.DataFrame(X_scaled, columns=features)

print("Scaled Feature Statistics (Mean ~ 0, Std ~ 1):")
df_scaled.describe().round(2).T[['mean', 'std', 'min', 'max']]
```

---
## 2. Storing Preprocessed Data
We save the standardized features so subsequent notebooks can load them directly.


```
scaled_path = Path("../data/processed/scaled_features.csv")
df_scaled.to_csv(scaled_path, index=False)
print(f"Scaled features saved to: {scaled_path.resolve()}")
```
