## 📌 Beginner Objective
Machine learning algorithms calculate geometric distances (Euclidean distance) between points.
- If borrower A makes **$80,000** and borrower B makes **$50,000**, the difference is **30,000**.
- But if their DTI differs by **5%**, the difference is only **5**.
- Without scaling, the algorithm would treat income as 6,000 times more important than DTI!

To fix this, we use **StandardScaler**:
$$\text{Scaled Value} = \frac{\text{Value} - \text{Mean}}{\text{Standard Deviation}}$$
This puts every variable on the exact same footing (Mean = 0, Std = 1).

> [!IMPORTANT]
> **Golden Rule of Unsupervised Clustering:**
> We **exclude** the `Default` column from training! We want our model to discover natural borrower behavior on its own.


```
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from pathlib import Path

df = pd.read_csv("../data/processed/cleaned_loans_dataset.csv")

# 1. Select ONLY behavioral features for clustering (exclude target 'Default')
features = ['annual_income', 'dti', 'loan_amount', 'fico_score', 'emp_length_years']
X = df[features].copy()

print("Features selected for clustering:")
print(X.head())
```

---
## 2. Apply StandardScaler


```
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert scaled numpy array back into a friendly DataFrame
df_scaled = pd.DataFrame(X_scaled, columns=features)

print("Scaled Feature Statistics (Notice Mean ~ 0 and Std ~ 1):")
df_scaled.describe().round(2).T[['mean', 'std', 'min', 'max']]
```

---
## 3. Save Preprocessed Scaled Features
We save the scaled data so subsequent notebooks can load it directly without repeating preprocessing.


```
scaled_path = Path("../data/processed/scaled_features.csv")
df_scaled.to_csv(scaled_path, index=False)
print(f"Saved scaled features to: {scaled_path.resolve()}")
```
