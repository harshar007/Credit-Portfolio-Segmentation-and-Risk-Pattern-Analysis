## Hierarchical Tree Analysis
Unlike K-Means, which requires picking $k$ in advance, **Agglomerative Hierarchical Clustering** works bottom-up:
1. Each borrower begins in their own individual cluster.
2. The closest pairs merge sequentially.
3. This creates a dendrogram showing natural hierarchy and separation thresholds.


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering

X_scaled = pd.read_csv("../data/processed/scaled_features.csv")

# We draw a representative sample of 250 borrowers so the dendrogram branches remain legible
np.random.seed(42)
sample_idx = np.random.choice(len(X_scaled), size=250, replace=False)
X_sample = X_scaled.iloc[sample_idx]

print(f"Sampled {len(X_sample)} borrowers for dendrogram analysis.")
```

---
## 1. Computing Linkage & Plotting the Dendrogram
Using Ward's method minimizes total within-cluster variance at each merge.


```
linkage_matrix = linkage(X_sample, method='ward')

plt.figure(figsize=(12, 6))
dendrogram(linkage_matrix, truncate_mode='level', p=5, no_labels=True, color_threshold=15)
plt.title("Hierarchical Clustering Dendrogram (Ward Linkage)")
plt.xlabel("Borrower Clusters")
plt.ylabel("Euclidean Distance (Dissimilarity)")
plt.axhline(y=15, color='red', linestyle='--', label='Cutoff threshold (k=4 clusters)')
plt.legend()

plt.tight_layout()
plt.savefig("../outputs/figures/hierarchical_dendrogram.png", dpi=300)
plt.show()
```

---
## 2. Fitting Agglomerative Clustering ($k=4$)


```
agg = AgglomerativeClustering(n_clusters=4, linkage='ward')
cluster_labels = agg.fit_predict(X_sample)

print("Hierarchical Cluster counts in sample:")
print(pd.Series(cluster_labels).value_counts().sort_index())
```
