## 📌 Beginner Objective
**Hierarchical Clustering (Agglomerative)** builds clusters bottom-up:
1. Every borrower starts in their own individual cluster.
2. The two most similar borrowers merge together.
3. This process repeats until all borrowers form a single tree called a **Dendrogram**.
4. We can visually inspect the dendrogram to see natural groupings without guessing $k$ in advance.


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering

# Load scaled features
X_scaled = pd.read_csv("../data/processed/scaled_features.csv")

# For dendrogram plotting, we take a representative sample of 250 borrowers
# (drawing 25,000 tree branches would look like solid black ink!)
np.random.seed(42)
sample_idx = np.random.choice(len(X_scaled), size=250, replace=False)
X_sample = X_scaled.iloc[sample_idx]

print(f"Sampled {len(X_sample)} borrowers for clean dendrogram visualization.")
```

---
## 1. Compute Linkage & Plot Dendrogram
We use **Ward's method**, which minimizes the variance when merging clusters.


```
linkage_matrix = linkage(X_sample, method='ward')

plt.figure(figsize=(12, 6))
dendrogram(linkage_matrix, truncate_mode='level', p=5, no_labels=True, color_threshold=15)
plt.title("Hierarchical Clustering Dendrogram (Ward Linkage)")
plt.xlabel("Borrower Clusters")
plt.ylabel("Euclidean Distance (Dissimilarity)")
plt.axhline(y=15, color='red', linestyle='--', label='Cutoff threshold (k=4 clusters)')
plt.legend()

plt.tight_layout()
plt.savefig("../outputs/figures/hierarchical_dendrogram.png", dpi=300)
plt.show()
```

---
## 2. Fit Agglomerative Clustering with $k=4$


```
agg = AgglomerativeClustering(n_clusters=4, linkage='ward')
cluster_labels = agg.fit_predict(X_sample)

print("Hierarchical Cluster counts in sample:")
print(pd.Series(cluster_labels).value_counts().sort_index())
```
