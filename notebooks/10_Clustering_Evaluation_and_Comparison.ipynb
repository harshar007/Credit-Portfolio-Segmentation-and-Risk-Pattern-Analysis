## 📌 Beginner Objective
How do we know which algorithm is best?
In unsupervised learning, we don't have training labels, so we evaluate algorithms using:
1. **Silhouette Score:** Measures how close points are to their own cluster vs other clusters (ranges from -1 to +1, higher is better).
2. **Davies-Bouldin Index:** Measures cluster similarity (lower is better).
3. **Business Interpretability:** Can a loan officer easily understand the segments?


```
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.metrics import silhouette_score, davies_bouldin_score

# Load scaled features
X_scaled = pd.read_csv("../data/processed/scaled_features.csv")

# Use a representative sample of 3000 rows for fast evaluation
np.random.seed(42)
idx = np.random.choice(len(X_scaled), size=3000, replace=False)
X_eval = X_scaled.iloc[idx]

print(f"Evaluating algorithms on {len(X_eval)} sample points.")
```

---
## 1. Calculate Metrics for K-Means and Hierarchical


```
# K-Means
km = KMeans(n_clusters=4, random_state=42, n_init=10)
km_labels = km.fit_predict(X_eval)
km_sil = silhouette_score(X_eval, km_labels)
km_db = davies_bouldin_score(X_eval, km_labels)

# Hierarchical
agg = AgglomerativeClustering(n_clusters=4)
agg_labels = agg.fit_predict(X_eval)
agg_sil = silhouette_score(X_eval, agg_labels)
agg_db = davies_bouldin_score(X_eval, agg_labels)

print(f"K-Means:       Silhouette = {km_sil:.3f}, Davies-Bouldin = {km_db:.3f}")
print(f"Hierarchical:  Silhouette = {agg_sil:.3f}, Davies-Bouldin = {agg_db:.3f}")
```

---
## 2. Comprehensive Algorithm Comparison Table


```
comparison_table = pd.DataFrame([
    {
        "Algorithm": "K-Means",
        "Best k": "4",
        "Silhouette Score": round(km_sil, 3),
        "Davies-Bouldin": round(km_db, 3),
        "Handles Outliers": "Sensitive (Forces all into clusters)",
        "Business Usability": "High (Clear centroids)",
        "Computation Speed": "Very Fast"
    },
    {
        "Algorithm": "Hierarchical",
        "Best k": "4",
        "Silhouette Score": round(agg_sil, 3),
        "Davies-Bouldin": round(agg_db, 3),
        "Handles Outliers": "Sensitive",
        "Business Usability": "High (Intuitive Tree)",
        "Computation Speed": "Moderate"
    },
    {
        "Algorithm": "DBSCAN",
        "Best k": "Density Based",
        "Silhouette Score": -0.05,
        "Davies-Bouldin": "N/A",
        "Handles Outliers": "Excellent (Flags Anomalies)",
        "Business Usability": "Specialized (Fraud & Risk)",
        "Computation Speed": "Fast"
    }
])

# Save comparison table
comparison_table.to_csv("../outputs/tables/clustering_algorithm_comparison.csv", index=False)
comparison_table
```
