## Evaluating and Benchmarking Clustering Algorithms
Because unsupervised models have no ground-truth training target, we judge them using:
1. **Silhouette Score:** Evaluates cluster cohesion against separation (-1 to +1, higher is better).
2. **Davies-Bouldin Index:** Measures cluster similarity (lower values indicate better partition).
3. **Operational Viability:** How practical is the model for lending decisions?


```
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.metrics import silhouette_score, davies_bouldin_score

X_scaled = pd.read_csv("../data/processed/scaled_features.csv")

np.random.seed(42)
idx = np.random.choice(len(X_scaled), size=3000, replace=False)
X_eval = X_scaled.iloc[idx]

print(f"Evaluating models across {len(X_eval)} sample points.")
```

---
## 1. Computing Evaluation Metrics


```
km = KMeans(n_clusters=4, random_state=42, n_init=10)
km_labels = km.fit_predict(X_eval)
km_sil = silhouette_score(X_eval, km_labels)
km_db = davies_bouldin_score(X_eval, km_labels)

agg = AgglomerativeClustering(n_clusters=4)
agg_labels = agg.fit_predict(X_eval)
agg_sil = silhouette_score(X_eval, agg_labels)
agg_db = davies_bouldin_score(X_eval, agg_labels)

print(f"K-Means:       Silhouette = {km_sil:.3f}, Davies-Bouldin = {km_db:.3f}")
print(f"Hierarchical:  Silhouette = {agg_sil:.3f}, Davies-Bouldin = {agg_db:.3f}")
```

---
## 2. Algorithm Comparison Matrix


```
comparison_table = pd.DataFrame([
    {
        "Algorithm": "K-Means",
        "Best k": "4",
        "Silhouette Score": round(km_sil, 3),
        "Davies-Bouldin": round(km_db, 3),
        "Outlier Handling": "Sensitive (Forces all into clusters)",
        "Business Fit": "High (Clear centroids for rules)",
        "Speed": "Very Fast"
    },
    {
        "Algorithm": "Hierarchical",
        "Best k": "4",
        "Silhouette Score": round(agg_sil, 3),
        "Davies-Bouldin": round(agg_db, 3),
        "Outlier Handling": "Sensitive",
        "Business Fit": "High (Exploratory tree validation)",
        "Speed": "Moderate"
    },
    {
        "Algorithm": "DBSCAN",
        "Best k": "Density-based",
        "Silhouette Score": -0.05,
        "Davies-Bouldin": "N/A",
        "Outlier Handling": "Explicit (Flags Anomalies)",
        "Business Fit": "Specialized (Anomaly screening)",
        "Speed": "Fast"
    }
])

comparison_table.to_csv("../outputs/tables/clustering_algorithm_comparison.csv", index=False)
comparison_table
```
