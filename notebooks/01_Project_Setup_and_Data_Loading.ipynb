## What We Are Solving Here
Financial institutions manage loans across borrowers with widely different income levels, credit scores, debt obligations, and employment backgrounds. Applying one standard policy across everyone is risky and leaves money on the table.

Our goal in this project is to group borrowers by their financial behavior using unsupervised machine learning, and then examine how those behavioral groups correlate with loan default rates.

In this first step, we will:
1. Set up our environment and import our core data science tools (`pandas`, `numpy`, `pathlib`).
2. Safely locate and load the raw loan file (`loans_granting_model_dataset.csv`).
3. Verify the records, review the columns, and confirm everything is ready for analysis.


```python
import os
from pathlib import Path
import pandas as pd
import numpy as np

print("Core libraries imported successfully!")
print(f"Pandas version: {pd.__version__}")
```

## Loading the Raw Loan Portfolio
We read the raw dataset from our data directory. We start with a 25,000-borrower slice so our workflow stays fast and responsive while we explore.


```python
data_path = Path("../data/raw/loans_granting_model_dataset.csv")

if not data_path.exists():
    data_path = Path("data/raw/loans_granting_model_dataset.csv")

print(f"Loading data from: {data_path.resolve()}")

df_raw = pd.read_csv(data_path, nrows=25000)
print(f"Data loaded successfully: {df_raw.shape[0]:,} rows and {df_raw.shape[1]} columns.")
```
