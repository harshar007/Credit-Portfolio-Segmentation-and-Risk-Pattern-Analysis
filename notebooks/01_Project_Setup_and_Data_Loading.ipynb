## 📂 Step 2: Define File Paths and Load Raw Data
In data science projects, keeping clean directory paths prevents "File Not Found" errors.
We read the raw CSV file located in `../data/raw/loans_granting_model_dataset.csv`.


```
# Define the relative path to our raw dataset
data_path = Path("../data/raw/loans_granting_model_dataset.csv")

if not data_path.exists():
    print(f"Warning: {data_path} not found directly. Checking local directory...")
    data_path = Path("data/raw/loans_granting_model_dataset.csv")

print(f"Loading data from: {data_path.resolve()}")

# Load first 25,000 records for fast, beginner-friendly exploration
df_raw = pd.read_csv(data_path, nrows=25000)
print(f"Data loaded successfully! Shape: {df_raw.shape[0]} rows, {df_raw.shape[1]} columns.")
```
