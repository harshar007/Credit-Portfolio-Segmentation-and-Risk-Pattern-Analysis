## Interactive Risk Analytics Interface
To bring our clustering models and portfolio insights into practical use for credit analysts, we packaged the system into an interactive Streamlit application.

Because Streamlit operates as an active web server, launch it from your terminal or command prompt with:
\\ash
streamlit run dashboard/app.py
\
This notebook verifies that the underlying data pipeline is ready and outlines the dashboard components.


```
import pandas as pd
from pathlib import Path

data_path = Path('../data/processed/segmented_loans_dataset.csv')
if not data_path.exists():
    data_path = Path('data/processed/segmented_loans_dataset.csv')

df = pd.read_csv(data_path)
print(f'Dashboard data ready: {len(df):,} loans loaded.')
df[['annual_income', 'fico_score', 'loan_amount', 'Segment_Name', 'Default']].head(5)
```

---
## Launching the Web Dashboard

Run the following command in your terminal:
\\ash
streamlit run dashboard/app.py
\
### Dashboard Capabilities:
1. **Portfolio KPI Dashboard:** Live tracking of loan counts, portfolio average FICO score, and aggregate default rates.
2. **Interactive 2D Cluster Visualizer:** Inspect individual borrower profiles in PCA space with dynamic Plotly tooltips.
3. **Segment Risk Summary:** Compare default rates, debt levels, and loan sizes across groups.
4. **Underwriting Risk Simulator:** Enter prospective applicant numbers (income, FICO, requested borrowing) to receive an instant risk assessment.
