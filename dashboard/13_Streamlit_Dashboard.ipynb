## 📌 Beginner Objective
Streamlit is an open-source Python framework that allows you to turn data models into interactive web applications without writing HTML, CSS, or JavaScript!

Because Streamlit runs as a continuous server, it is executed from the terminal using:
\\ash
streamlit run dashboard/app.py
\
In this notebook, we preview the dashboard architecture, inspect its backend data, and show how to launch it.


```
import pandas as pd
from pathlib import Path

# Verify that the segmented data exists for the dashboard
data_path = Path('../data/processed/segmented_loans_dataset.csv')
if not data_path.exists():
    data_path = Path('data/processed/segmented_loans_dataset.csv')

df = pd.read_csv(data_path)
print(f'Dashboard data ready! Total loans: {len(df):,}')
df[['annual_income', 'fico_score', 'loan_amount', 'Segment_Name', 'Default']].head(5)
```

---
## 🚀 How to Launch the Streamlit Web Dashboard

Open your terminal or PowerShell and run:
\\ash
streamlit run dashboard/app.py
\This opens an interactive dashboard in your web browser with:
1. **Executive Portfolio KPIs:** Total volume, Avg FICO, Portfolio default rate.
2. **2D Interactive Cluster Plot:** Zoom in and inspect individual borrowers in PCA space using Plotly.
3. **Risk Breakdown Table:** Compare default rates and financial characteristics across segments.
4. **Interactive Underwriter Risk Simulator:** Type in any applicant's income and FICO score to get an instant risk assessment!
